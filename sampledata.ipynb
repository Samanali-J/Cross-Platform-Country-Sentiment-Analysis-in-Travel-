```python
import json
import random
import pandas as pd
```

```python
youtube = pd.read_csv('Datasets/yt_comments.csv')
sample_yt = youtube.sample(n=50, random_state=42)
sample_yt.to_csv('youtube_sample_dataset.csv', index=False)
```

```python
with open('Datasets/combined_unique_reddit_posts.json', 'r') as f:
    data = json.load(f)
    data = data['submissions']

sample_data = random.sample(data, 50)  

with open('reddit_sample_dataset.json', 'w') as f:
    json.dump(sample_data, f, indent=2)
```
